```python
import pandas as pd
import io

# 1. Create a CSV with no header and 5 columns
csv_data = """
1,Alice,Math,95,A
2,Bob,Science,88,B
3,Charlie,History,92,A
4,David,Art,85,B
5,Eve,Music,98,A
"""

# Load it with our own column names and make the first column the index
column_names = ['id', 'name', 'subject', 'score', 'grade']
df = pd.read_csv(io.StringIO(csv_data),
                 header=None,
                 names=column_names,
                 index_col=0)

print(df)
```

```python
import pandas as pd
import numpy as np
import tempfile
import os

# Create a sample DataFrame with missing values
data = {
    'A': [1, 2, np.nan, 4],
    'B': ['foo', 'bar', 'baz', None]
}
df_to_write = pd.DataFrame(data)

# Use a temporary file to demonstrate writing and reading
with tempfile.NamedTemporaryFile(delete=False,
                                 suffix='.csv',
                                 mode='w') as tmp:
    temp_filename = tmp.name
    # 2. Write file using ; as delimiter and -999 for missing values
    df_to_write.to_csv(temp_filename,
                       sep=';',
                       na_rep='-999',
                       index=False)

print("Written CSV content:")
with open(temp_filename, 'r') as f:
    print(f.read())

# Read it back so -999 becomes NaN
df_read = pd.read_csv(temp_filename,
                      sep=';',
                      na_values='-999')
print("Read DataFrame:")
display(df_read)

# Clean up the temporary file
os.remove(temp_filename)
```

```python
import pandas as pd

# Path to the titanic CSV
file_path = '/content/titanic.csv'
'''
We will compute the mean of the 'Fare'
column grouped by 'Pclass' (passenger class)
Since we are processing in chunks, we aggregate
the sum and count per class for each chunk

'''
sums = pd.Series(dtype=float)
counts = pd.Series(dtype=int)

'''
Read in chunks of 2,000 rows (titanic.csv is small,
but this demonstrates the chunk logic)

'''
chunksize = 2000
for chunk in pd.read_csv(file_path, chunksize=chunksize):
    # Group by Pclass and aggregate sum and count of Fare
    grouped = chunk.groupby('Pclass')['Fare'].agg(['sum', 'count'])

    # Update global running sums and counts
    sums = sums.add(grouped['sum'], fill_value=0)
    counts = counts.add(grouped['count'], fill_value=0)

# Compute final mean across all chunks
final_mean = sums / counts
print("Mean of 'Fare' per 'Pclass' across all chunks:")
print(final_mean)
```

```python
import pandas as pd

# Create a sample DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'city': ['New York', 'Paris', 'London']
}
df_original = pd.DataFrame(data)

# 4. Export the DataFrame with orient="records"
json_str = df_original.to_json(orient="records")
print("Exported JSON string:")
print(json_str)
print()

# Read it back with pd.read_json (since it is orient="records", we specify the orient)
df_restored = pd.read_json(json_str, orient="records")
print("Restored DataFrame:")
print(df_restored)
print()

# Check equality with .equals
are_equal = df_original.equals(df_restored)
print(f"Are the original and restored DataFrames equal? {are_equal}")
```

```python
import pandas as pd
import io

# 5. Write a short HTML table by hand
html_table = """
<table>
  <thead>
    <tr>
      <th>Product</th>
      <th>Price</th>
      <th>In Stock</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>Laptop</td>
      <td>$999</td>
      <td>Yes</td>
    </tr>
    <tr>
      <td>Phone</td>
      <td>$499</td>
      <td>No</td>
    </tr>
  </tbody>
</table>
"""

# Load it with read_html
# pd.read_html returns a list of DataFrames (one for each table found in the HTML)
dfs = pd.read_html(io.StringIO(html_table))
df_html = dfs[0]

print("Loaded HTML table as a DataFrame:")
display(df_html)
```
